```python
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.stats import norm
from scipy.optimize import brentq
import sys; sys.path.insert(0,'../src')
import warnings; warnings.filterwarnings('ignore')
np.random.seed(42)

plt.rcParams.update({
    'figure.facecolor':'#0d1117','axes.facecolor':'#0d1117',
    'axes.edgecolor':'#30363d','axes.labelcolor':'#c9d1d9',
    'xtick.color':'#8b949e','ytick.color':'#8b949e',
    'text.color':'#c9d1d9','grid.color':'#21262d',
    'axes.titlecolor':'#f0f6fc','legend.facecolor':'#161b22',
    'legend.edgecolor':'#30363d','font.size':11
})
COLORS = ['#58a6ff','#3fb950','#f78166','#d2a8ff','#ffa657','#79c0ff']
returns = pd.read_csv('../data/returns.csv', index_col='Date', parse_dates=True)
spy_ret = returns['SPY']
S0, r, sigma = 100, 0.05, 0.20
print(f'Loaded {len(spy_ret):,} observations')
```

```python
from implied_vol import vol_surface_model, build_vol_surface
strikes_s=np.linspace(75,130,40); mats_s=[0.25,0.5,1.0,2.0]; lbls=["3M","6M","1Y","2Y"]
fig,axes=plt.subplots(1,2,figsize=(15,6),facecolor="#0d1117")
for T,lbl,c in zip(mats_s,lbls,COLORS):
    smile=vol_surface_model(strikes_s,S0,T)*100
    axes[0].plot(strikes_s,smile,color=c,linewidth=2.5,label=lbl,marker="o",markersize=3)
axes[0].axvline(S0,color="#8b949e",linewidth=1.2,linestyle="--",alpha=0.7,label="ATM")
axes[0].set_title("Volatility Smile",color="#f0f6fc")
axes[0].set_xlabel("Strike",color="#8b949e"); axes[0].set_ylabel("Implied Vol (%)",color="#8b949e")
axes[0].legend(); axes[0].grid(True,alpha=0.3); axes[0].tick_params(colors="#8b949e")
m_ax=np.linspace(-2,2,100)
for T,lbl,c in zip(mats_s,lbls,COLORS):
    vol=(0.20+0.03*np.exp(-T))+(-0.12/np.sqrt(T))*m_ax+(0.06/T)*m_ax**2
    axes[1].plot(m_ax,vol*100,color=c,linewidth=2.5,label=lbl)
axes[1].axvline(0,color="#8b949e",linewidth=1.2,linestyle="--",alpha=0.7)
axes[1].set_title("Smile by Log-Moneyness",color="#f0f6fc")
axes[1].set_xlabel("Log-Moneyness",color="#8b949e"); axes[1].legend(); axes[1].grid(True,alpha=0.3)
axes[1].tick_params(colors="#8b949e")
plt.tight_layout()
plt.savefig("../results/04_volatility_smile.png",dpi=150,bbox_inches="tight",facecolor="#0d1117")
plt.show()
print("Chart saved.")
```
