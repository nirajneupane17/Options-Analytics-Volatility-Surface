```python
from black_scholes import bs_call, bs_put, pricing_table, put_call_parity
strikes=[80,90,95,100,105,110,120]; mats=[0.25,0.5,1.0,2.0]
df=pricing_table(S0,r,sigma,strikes,mats)
print(df[df.Maturity==1.0].to_string(index=False))
df.to_csv('../results/options_pricing_summary.csv',index=False)
```

```python
# BS Pricing Surface 3D
from black_scholes import bs_call
K_arr=np.linspace(70,135,50); T_arr=np.linspace(0.05,2,50)
Kg,Tg=np.meshgrid(K_arr,T_arr)
Cg=np.vectorize(lambda K,T: bs_call(S0,K,T,r,sigma))(Kg,Tg)
fig=plt.figure(figsize=(13,8),facecolor="#0d1117")
ax=fig.add_subplot(111,projection="3d",facecolor="#0d1117")
surf=ax.plot_surface(Kg,Tg,Cg,cmap="plasma",alpha=0.92,linewidth=0,antialiased=True)
ax.set_xlabel("Strike",color="#8b949e",labelpad=10)
ax.set_ylabel("Maturity",color="#8b949e",labelpad=10)
ax.set_zlabel("Call Price",color="#8b949e",labelpad=10)
ax.set_title("Black-Scholes Call Price Surface",color="#f0f6fc",pad=15)
ax.tick_params(colors="#8b949e")
ax.xaxis.pane.fill=ax.yaxis.pane.fill=ax.zaxis.pane.fill=False
ax.view_init(elev=28,azim=-55)
fig.colorbar(surf,ax=ax,shrink=0.5,aspect=10)
plt.tight_layout()
plt.savefig("../results/01_bs_pricing_surface.png",dpi=150,bbox_inches="tight",facecolor="#0d1117")
plt.show()
print("Chart saved.")
```
